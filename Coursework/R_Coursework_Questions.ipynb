### Question 10

Sticking with the `diamonds` dataset for this question. A carat is a special unit of weight for gemstones where a single carat is 200 mg or 0.2 g. Write a function to convert the `diamonds$carat` vector into grams or milligrams. How heavy is the heaviest diamond?

```r
library(ggplot2) #load required package
data("diamonds") #load data

convert_carat_to_mg <- function(x) { #function name
  mg <- (x * 200) #numerical formula for conversion
  return(mg)
}

heaviest_mg <- convert_carat_to_mg(diamonds$carat) #converts all carats to mg and creates a max result name
max(heaviest_mg) #print the heaviest diamond
```
