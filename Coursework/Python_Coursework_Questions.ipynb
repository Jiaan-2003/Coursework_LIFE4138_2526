### Question 3

Use three different methods to print the numbers 200 to 300.

````python
for i in  range (200, 301): #creates a for loop
    print(i) #prints numbers
````

````python
x = 200 #start from this number
while x <= 300: #sets loop to continue while x is less than 300
    print(x) # prints numbers
    x += 1 #adds 1 on to x for each number
````

````python
nums = [i for i in range (200, 301)] #creates a list of numbers in a range

for num in nums: #loops the output
    print(num) #prints the numbers
````

### Question 5
Using pandas, generate a scatterplot of mpg vs hp using the cars dataset. The data set is read in and assigned to a dataframe by running the cell below. Ensure you have imported all the libraries you need. 

````python
mtcars = "https://raw.githubusercontent.com/Apress/mastering-ml-w-python-in-six-steps/refs/heads/master/Chapter_2_Code/Data/mtcars.csv"
cars = pd.read_csv(mtcars)
````

### Question 6
Using the startwars dataset imported below, identify the individuals with tallest height and lowest mass.

````python
import pandas as pd #imports pandas
starwars_dataset="https://www.fabricionarcizo.com/post/starwars/updated_starwars.csv"
starwars = pd.read_csv(starwars_dataset)
````

````python
Tallest = starwars.sort_values(by = "height", ascending=False).head(1) #sorts the data to print only the tallest individual in the dataset by looking at the height data only
print(Tallest) #prints the tallest individuals information
````

````python
Lowest_mass = starwars.sort_values(by = "mass", ascending=True).head(1) #sorts the data to print only the lowest mass individual in the dataset by looking at the mass data only
print(Lowest_mass) #prints the information of the individual with the lowest mass
````
