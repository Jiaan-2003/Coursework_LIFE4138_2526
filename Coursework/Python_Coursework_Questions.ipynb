### Question 3

Use three different methods to print the numbers 200 to 300.

````python
for i in  range (200, 301): #creates a for loop
    print(i) #prints numbers
````

````python
x = 200 #start from this number
while x <= 300: #sets loop to continue while x is less than 300
    print(x) # prints numbers
    x += 1 #adds 1 on to x for each number
````

````python
nums = [i for i in range (200, 301)] #creates a list of numbers in a range

for num in nums: #loops the output
    print(num) #prints the numbers
````

````python
import seaborn as sns #imports seaborn
import matplotlib.pyplot as plt #imports matplotlib
%matplotlib inline  
diamonds = sns.load_dataset("diamonds") #loads data

sns.scatterplot(data = diamonds, x = "carat", y = "price", #applies the required data and sets the axis labels
                color="blue", #mkaes the plots blue
                alpha = 0.5,) #customises plot transparency (partially transparent)
plt.xlabel("Carat") #labels x axis
plt.ylabel("Price") #labels y axis
plt.title("Carat vs Price") #titles the plot
plt.show() #prints the scatter plot
````

````python
import seaborn as sns #imports seaborn
import matplotlib.pyplot as plt #imports matplotlib
%matplotlib inline  
diamonds = sns.load_dataset("diamonds") #loads data

sns.histplot(diamonds["price"], bins=30, color="green") #applies the required data, sizes the bars and colours them green
plt.xlabel("Price") #labels the x axis
plt.ylabel("Count") #labels the y axis
plt.title("Histogram of Diamond Prices") #gives the graph a title
plt.show() #plots the histogram
````

````python
import seaborn as sns #imports seaborn
import matplotlib.pyplot as plt #imports matplotlib
%matplotlib inline  
diamonds = sns.load_dataset("diamonds") #loads data

sns.boxplot(data = diamonds, x = "cut", y = "price" , color = "red", linewidth = 2) #selects data for the plot, selects data for the axis, makes the boxplots red and sets the width of boxplot lines. 
plt.title("Boxplot of Diamond Cut Against Prices") #titles the boxplot 
plt.xlabel("Cut") #labels the x axis
plt.ylabel("Price") #labels the y axis
plt.show() #displays the plot
````

### Question 9

Using the same dataset, identify the max and min price in diamonds dataset.

````python
import seaborn as sns #imports seaborn
import matplotlib.pyplot as plt #imports matplotlib
%matplotlib inline  
import numpy as np #imports numpy
diamonds = sns.load_dataset("diamonds") #loads data

min_price = diamonds["price"].min() #pulls the minimum price diamond from the dataset
max_price = diamonds["price"].max() #pulls the maximum price diamond from the dataset

diamonds2 = diamonds.copy() #copies the dataset into a new diamonds2 copy
diamonds2["Price Band"] = np.where(diamonds2["price"] < 5000, #uses the numpy where function to create a new column for the lowest and highest price bands
    "Low Price", "High Price") #the two price band labels selected

print(max_price) #prints the maximum price diamond
print(min_price) #prints the minimum price diamond
````

### Question 10
Write a function that will convert carat to grams and find the heaviest diamond.

1 carat = 0.2 grams

````python
import seaborn as sns #imports seaborn
diamonds = sns.load_dataset("diamonds") #loads data

def convert_carat_to_mg(x): #defines a function to convert carat to mg
    mg = x *  200 #numerical formula for conversion
    return mg

heaviest_mg = convert_carat_to_mg(diamonds["carat"]) #converts all carats to mg and creates a max result name

print(max(heaviest_mg)) #prints the heavist diamond 
    

````
